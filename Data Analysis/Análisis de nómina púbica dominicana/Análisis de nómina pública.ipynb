# Análisis de las nóminas de ITLA, MESCyT y MINPRE personal fijo Nov 2025

## Importaciones

```python
import pandas as pd
```

## Carga y limpieza de ITLA

```python
df_itla = pd.read_excel("itla_2025_11_fijos.xlsx", skiprows=8)
df_itla
```

```python
# Eliminar filas nulas
df_itla = df_itla[~df_itla["Salario Bruto"].isna()]
df_itla
```

```python
# Eliminar columna innecesaria
df_itla = df_itla.drop("Unnamed: 11", axis=1)
df_itla
```

```python
# Renombrar columnas
mapping_itla = {
    "Servidor Público": "Nombre completo",
    "Sexo": "Género",
    "Cargo": "Función",
    "Neto": "Salario Neto"
}
df_itla = df_itla.rename(columns=mapping_itla)
df_itla
```

```python
# Agregar columnas informativas
df_itla["Institución"] = "ITLA"
df_itla["Año"] = 2025
df_itla["Mes"] = "Noviembre"
df_itla
```

```python
df_itla.info()
```

## Carga y limpieza de MESCyT

```python
df_mescyt = pd.read_excel("mescyt_2025_11_fijos.xlsx", skiprows=10)
df_mescyt
```

```python
df_mescyt = df_mescyt.iloc[:-1]
df_mescyt
```

```python
# Renombrar columnas
mapping_mescyt = {
    "Nombre y Apellidos": "Nombre completo",
    "Genero": "Género",
    "Dirección-Departamento": "Departamento",
    "Sueldo Bruto": "Salario Bruto",
    "Sueldo Neto": "Salario Neto",
    "Total \nDesc.": "Total Desc."
}
df_mescyt = df_mescyt.rename(columns=mapping_mescyt)
df_mescyt
```

```python
# Agregar columnas informativas
df_mescyt["Institución"] = "MESCyT"
df_mescyt["Año"] = 2025
df_mescyt["Mes"] = "Noviembre"
df_mescyt
```

```python
df_mescyt.info()
```

## Carga y limpieza de MINPRE

```python
df_minpre = pd.read_excel("minpre_2025_11_fijos.xlsx", skiprows=4)
df_minpre
```

```python
# Eliminar columnas que no nos interesan
df_minpre = df_minpre.drop(["Unnamed: 7", "Unnamed: 10"], axis=1)
df_minpre
```

```python
# Agregar encabezados concatenados correctos
combined_headers = df_minpre.iloc[:3].bfill(axis=0).iloc[0,:].tolist()
df_minpre.columns = combined_headers
df_minpre = df_minpre.iloc[3:]
df_minpre
```

```python
# Eliminar filas sin datos de importancia para el análisis
df_minpre = df_minpre[~df_minpre["ESTATUS "].isna()]
df_minpre
```

```python
# Renombrar columnas
mapping_minpre = {
    "NOMBRE": "Nombre completo",
    "FUNCION": "Función",
    "ESTATUS ": "Estatus",
    "GENERO": "Género",
    "SUELDO BRUTO\n(RD$)": "Salario Bruto",
    "SUELDO NETO ": "Salario Neto",
    "TOTAL DESCUENTOS": "Total Desc.",
    "IMPUESTO SOBRE LA RENTA": "ISR",
    "SEGURO FAMILIAR DE SALUD": "SFS",
    "SEGURIDAD SOCIAL LEY 87 01": "AFP"
}
df_minpre = df_minpre.rename(columns=mapping_minpre)
df_minpre
```

```python
# Agregar columnas informativas
df_minpre["Institución"] = "MINPRE"
df_minpre["Año"] = 2025
df_minpre["Mes"] = "Noviembre"
df_minpre
```

```python
df_minpre.info()
```

## Concatenación de todo

```python
df_nomina = pd.concat([df_itla, df_mescyt, df_minpre], axis=0)
df_nomina = df_nomina[[
    "Institución", "Año", "Mes", "Nombre completo", "Género", "Función", "Departamento", "Estatus",
    "Salario Bruto", "ISR", "SFS", "AFP", "Total Desc.", "Salario Neto"
]].reset_index(drop=True)
df_nomina
```

```python
numeric_columns = ["Salario Bruto", "ISR", "SFS", "AFP", "Total Desc.", "Salario Neto"]
for column in numeric_columns:
  df_nomina[numeric_columns] = df_nomina[numeric_columns].astype(float)
df_nomina
```

```python
df_nomina.info()
```

```python
# Hace falta homologar las columnas categóricas para que el análisis se pueda realizar correctamente. Aquí un ejemplo con la columna "Función"
df_nomina["Función"].value_counts()
```

```python
# ¿Cómo resolver ese problema? Veamos un ejemplo con las columnas Género y Estatus
```

```python
print("Género antes de homologar:", df_nomina["Género"].value_counts())

genero_mapping = {
    "MASCULINO": "M",
    "FEMENINO": "F"
}
df_nomina["Género"] = df_nomina["Género"].apply(lambda x: genero_mapping.get(x, x))

print("Género después de homologar:", df_nomina["Género"].value_counts())
```

```python
df_nomina["Estatus"].value_counts()
```

```python
print("Estatus antes de homologar:", df_nomina["Estatus"].value_counts())

def estatus_mapping(x):
  x = x.upper()
  if "CONFIANZA" in x:
    return "PERSONAL DE CONFIANZA"
  elif "LIBRE" in x:
    return "LIBRE NOMBRAMIENTO Y REMOCION"
  return x

df_nomina["Estatus"] = df_nomina["Estatus"].apply(estatus_mapping)

print("Estatus después de homologar:", df_nomina["Estatus"].value_counts())
```
