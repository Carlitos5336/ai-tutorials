## Concatenación general

```python
# Empaquetando toda esta lógica en una función para reproducirlo en los demás datasets
def clean_data(df):
  # Separar columna de duration
  df["n_minutes"] = df.apply(lambda x: x["duration"].split(" ")[0] if "min" in str(x["duration"]) else np.nan, axis=1).astype(float)
  df["n_seasons"] = df.apply(lambda x: x["duration"].split(" ")[0] if "Season" in str(x["duration"]) else np.nan, axis=1).astype(float)
  # prompt: Convierte la columna date_added de object a datetime. Por ejemplo: el string "July 1, 2019" a datetime
  df["date_added"] = pd.to_datetime(df["date_added"], format='%B %d, %Y', errors='coerce')
  # Agregar columna de años de antiguedad del show/película en la plataforma y publicado
  df["years_old_added"] = df["date_added"].apply(lambda x: datetime.datetime.now().year - x.year)
  df["years_old_published"] = df["release_year"].apply(lambda x: datetime.datetime.now().year - x)
  # Diferencia de años de publicación y agregado a la plataforma
  df["years_difference"] = df["years_old_published"] - df["years_old_added"]
  # Calcular el número de actores
  df["n_cast_members"] = df["cast"].apply(lambda x: len(str(x).split(",")) if str(x) != "nan" else 0)
  # Calculando la lista de géneros distintos
  lista_de_generos = df["listed_in"].apply(lambda x: [i.strip() for i in str(x).split(",")])
  lista_unica_de_generos = set()
  for i in lista_de_generos:
    lista_unica_de_generos = lista_unica_de_generos.union(set(i))
  # Agregar genero encoded
  for genero in lista_unica_de_generos:
    df[f"is_{genero}"] = lista_de_generos.apply(lambda x: True if genero in x else False)
  return df
```

```python
df_netflix = pd.read_csv("netflix_titles.csv")
df_disney_plus = pd.read_csv("disney_plus_titles.csv")
df_amazon_prime = pd.read_csv("amazon_prime_titles.csv")
```

```python
df_netflix = clean_data(df_netflix)
df_netflix["platform"] = "Netflix"

df_disney_plus = clean_data(df_disney_plus)
df_disney_plus["platform"] = "Disney Plus"

df_amazon_prime = clean_data(df_amazon_prime)
df_amazon_prime["platform"] = "Amazon Prime"
```

```python
df_full = pd.concat([df_netflix, df_disney_plus, df_amazon_prime]).reset_index(drop=True)
df_full
```

```python
# 1. Homologar con un diccionario
rating_mapping_dict = {
    "AGES_18_": "Adultos",
    "AGES_16_": "Adultos",
    "TV-14": "Niños",
    "TV-MA": "Niños"
}
df_full["rating"].apply(lambda x: rating_mapping_dict[x] if x in rating_mapping_dict else x).value_counts()
```

```python
# 2. Homologar con una función
def rating_mapping_func(x):
  x = str(x).lower()
  if "min" in x: return "Not Rated"
  elif "13" in x: return "Discretion"
  elif "16" in x: return "Discretion"
  elif "18" in x: return "Discretion"
  elif "14" in x: return "Discretion"
  elif "17" in x: return "Discretion"
  elif "-ma" in x: return "Discretion"
  elif "pg" in x: return "Discretion"
  elif "nr" in x: return "Not Rated"
  elif "nan" in x: return "Not Rated"
  elif "r" in x: return "Discretion"
  else: return "All Public"
df_full["rating"].apply(lambda x: rating_mapping_func(x)).value_counts()
```

```python
# Aplicando transformación real
print(f"Cantidad de categorías de Rating: {len(df_full['rating'].unique())}")
df_full["rating"] = df_full["rating"].apply(lambda x: rating_mapping_func(x))
print(f"Cantidad de categorías de Rating después de resumir: {len(df_full['rating'].unique())}")
```
